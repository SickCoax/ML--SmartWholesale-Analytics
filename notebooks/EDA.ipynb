 # EDA

```python
# Imports

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Data Load

df = pd.read_csv("../dataset/Wholesale customers data.csv")

df
```

```python
# Stats

print(df.describe())
```

```python
# Null Values

print(df.isnull().sum())
```

```python
# Channel Distribution

sns.countplot(
    x = "Channel" ,
    data = df
)
plt.title("Channel Distribution")
plt.grid()
plt.show()
```

# Channel 1 (Horeca) accounts for the majority of customers in the dataset,
# while Channel 2 (Retail) represents a smaller portion.
# This suggests the wholesale distributor serves more Horeca businesses than Retail businesses.

```python
# Region Distribution

sns.countplot(
    x = "Region" ,
    data = df
)
plt.title("Region Distribution")
plt.grid()
plt.show()
```

# This suggest the wholesale distributor serves more in Region 3

```python
# Heat Map

sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    cmap="coolwarm"
)

plt.title("Heat correlation")
plt.show()
```

# • Grocery and Detergents_Paper show a very strong positive correlation (0.92).
# • Milk is strongly correlated with Grocery (0.73) and Detergents_Paper (0.66).
# • Fresh and Frozen products exhibit a moderate positive relationship (0.35).
# • Region has little impact on customer purchasing behavior.
# • Channel is strongly associated with Grocery and Detergents_Paper spending, indicating distinct purchasing patterns between Horeca and Retail customers.
